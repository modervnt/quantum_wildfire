```python
# rdNBR

def fetch_rgb_nbr(bbox, size, time_interval):
    req = SentinelHubRequest(
        evalscript=evalscript,
        input_data=[
            SentinelHubRequest.input_data(
                data_collection = DataCollection.SENTINEL2_L2A,
                time_interval = time_interval,
                mosaicking_order = "leastCC"
            )
        ],
        responses = [SentinelHubRequest.output_response("default", MimeType.TIFF)],
        bbox = bbox,
        size = size,
        config = config
    )
    data = req.get_data()[0]
    return data[:,:,0:3], data[:,:,3]

#This streching function maakes the rgb picture more clear
def stretch(band, p_low=2, p_high=98):
    lo, hi = np.nanpercentile(band, [p_low, p_high])
    return np.clip((band - lo) / (hi - lo + 1e-6), 0, 1)

rgb_before, nbr_before = fetch_rgb_nbr(
    bbox2,size2, time_interval =("2026-06-01", "2026-06-01")
)
rgb_after, nbr_after = rgb_after, nbr_after = fetch_rgb_nbr(
    bbox2, size2, time_interval=("2026-09-22", "2026-09-22")
)
print("Shape before :", rgb_before.shape, "| Shape after :",rgb_after.shape)

dnbr = nbr_before - nbr_after
eps = 1e-6 #this is for avoid division by 0
rdnbr = dnbr / np.sqrt(np.abs(nbr_before) + eps)

fig, axes = plt.subplots(1,2, figsize=(20,10))

rgb_strechted = np.dstack([stretch(rgb_after[:,:,i]) for i in range(3)])
axes[0].imshow(rgb_strechted)
axes[0].set_title("Image RGB (After)")
axes[0].axis("off")


#for the rdnbr, the colormap is the inverse of the ndbr
im = axes[1].imshow(rdnbr, cmap="RdYlGn_r", vmin=-1, vmax=1)
axes[1].set_title("rdNBR ")
axes[1].axis("off")
plt.colorbar(im, ax=axes[1], fraction=0.046, pad=0.04)

plt.tight_layout()
plt.show()

# --- Statistiques rapides ---
print("rdNBR min :", np.nanmin(rdnbr))
print("rdNBR max :", np.nanmax(rdnbr))
print("rdNBR moyen :", np.nanmean(rdnbr))
```

```python
import pandas as pd
import numpy as np

bornes_nbr = [
    -0.99943787, -0.6542056, -0.16348448, 
    0.25706846, 0.5435323, 0.9910204
]

bornes_rdnbr = [
    -924.18604, -427.41934, -129.03224, 
    196.07843, 615.3846, 998.5337
]

classes_severite = ['Non brule', 'Faible', 'Modere', 'Severe', 'Tres fort']

nbr_1d = nbr.flatten()
rdnbr_1d = rdnbr.flatten()


if len(nbr_1d) != len(rdnbr_1d):
    raise ValueError(f"Erreur: nbr a {len(nbr_1d)} pixels mais rdnbr en a {len(rdnbr_1d)}.")

data = {
    'pixel_id': range(1, len(nbr_1d) + 1), 
    'nbr': nbr_1d,   
    'rdnbr': rdnbr_1d 
}

df = pd.DataFrame(data)

df['Classe_NBR'] = pd.cut(
    df['nbr'], 
    bins=bornes_nbr, 
    labels=classes_severite, 
    include_lowest=True
)

df['Classe_RDNBR'] = pd.cut(
    df['rdnbr'], 
    bins=bornes_rdnbr, 
    labels=classes_severite, 
    include_lowest=True
)

def determiner_severite_finale(row):
    if row['Classe_NBR'] == row['Classe_RDNBR']:
        return row['Classe_NBR']
    elif pd.isna(row['Classe_NBR']) or pd.isna(row['Classe_RDNBR']):
        return 'Hors limites'
    else:
        return 'Indetermine (Mixte)'

df['Severite_Finale'] = df.apply(determiner_severite_finale, axis=1)

print(df[['pixel_id', 'nbr', 'rdnbr', 'Classe_NBR', 'Classe_RDNBR', 'Severite_Finale']].head(10))
```
